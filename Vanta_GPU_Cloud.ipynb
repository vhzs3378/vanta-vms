```python
# @title 2. Iniciar Sistema Operacional com RDP e Túnel de Alta Velocidade (CLIQUE NO PLAY ▶️)
# @markdown Defina sua senha de acesso remoto:
SENHA_RDP = "Pass@Vanta2026!" # @param {type:"string"}

import subprocess, time, urllib.request, zipfile, os, re

print("[*] Configurando Área de Trabalho Remota com Aceleração de GPU NVIDIA...")

# 1. Instalar Desktop XFCE e servidor XRDP otimizado com suporte a GPU
subprocess.run(["apt-get", "update", "-qq"], check=True)
subprocess.run(["apt-get", "install", "-y", "-qq", "xfce4", "xfce4-goodies", "xorg", "dbus-x11", "x11-xserver-utils", "xrdp"], check=True)

# 2. Configurar usuário runneradmin
subprocess.run(["useradd", "-m", "-s", "/bin/bash", "runneradmin"], check=False)
subprocess.run(["sh", "-c", f"echo 'runneradmin:{SENHA_RDP}' | chpasswd"], check=True)
subprocess.run(["usermod", "-aG", "sudo", "runneradmin"], check=True)

# 3. Configurar sessão gráfica XFCE
with open("/home/runneradmin/.xsession", "w") as f:
    f.write("exec startxfce4\n")
subprocess.run(["chown", "runneradmin:runneradmin", "/home/runneradmin/.xsession"], check=True)
subprocess.run(["adduser", "xrdp", "ssl-cert"], check=False)

# 4. Iniciar serviço XRDP
subprocess.run(["service", "xrdp", "restart"], check=True)

# 5. Baixar e iniciar túnel Bore para expor porta 3389 na internet
print("[*] Inicializando túnel TCP ultrarrápido (Bore)...")
if not os.path.exists("bore"):
    urllib.request.urlretrieve("https://github.com/ekzhang/bore/releases/download/v0.6.0/bore-v0.6.0-x86_64-unknown-linux-musl.tar.gz", "bore.tar.gz")
    subprocess.run(["tar", "-xzf", "bore.tar.gz"], check=True)
    subprocess.run(["chmod", "+x", "bore"], check=True)

bore_proc = subprocess.Popen(["./bore", "local", "3389", "--to", "bore.pub"], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)

endpoint = ""
for _ in range(20):
    line = bore_proc.stdout.readline()
    if "listening at" in line:
        m = re.search(r"listening at (bore\.pub:\d+)", line)
        if m:
            endpoint = m.group(1)
            break
    time.sleep(0.5)

print("\n" + "="*60)
print("🎉 SUA MÁQUINA VIRTUAL COM GPU NVIDIA TESLA T4 ESTÁ ONLINE!")
print("="*60)
print(f"📌 COMPUTADOR / HOST:  {endpoint}")
print(f"👤 USUÁRIO:            runneradmin")
print(f"🔑 SENHA:              {SENHA_RDP}")
print("="*60)
print("💡 Como conectar: Abra o programa 'Conexão de Área de Trabalho Remota' no Windows,")
print(f"   cole '{endpoint}', usuário 'runneradmin' e senha '{SENHA_RDP}'!")
print("="*60 + "\n")

# Manter ativo
try:
    while True:
        time.sleep(60)
except KeyboardInterrupt:
    print("Máquina encerrada pelo usuário.")
```
